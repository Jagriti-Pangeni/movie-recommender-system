```python
import pandas as pd
import numpy as np
```

```python
movies=pd.read_csv("tmdb_5000_movies.csv")
credits=pd.read_csv("tmdb_5000_credits.csv")
```

```python
movies.head()
movies=movies.merge(credits,on="title")
movies.columns
```

```python
# creating a working dataset
movies=movies[["movie_id","title","overview","genres","keywords","cast","crew"]]
```

```python
movies.isnull().sum()
```

```python
# movies.dropna(inplace=True)
movies.duplicated(subset=["title","movie_id"]).sum()
```

```python
# Look at the rows where title+movie_id match but the full row doesn't
# dupes_by_key = movies[movies.duplicated(subset=["title","movie_id"], keep=False)]
# dupes_by_key.sort_values(["title","movie_id"])
```

```python
movies = movies.drop_duplicates(subset=["movie_id"], keep="first")
movies = movies.reset_index(drop=True)
```

```python
movies["genres"]
```

```python
import ast
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i['name'])
    return L
movies['genres']=movies['genres'].apply(convert)
```

```python
movies.head()
```

```python
movies["keywords"]=movies["keywords"].apply(convert)
```

```python
def convert3(obj):
    L=[]
    count=0
    for i in ast.literal_eval(obj):
        if count!=3:
            L.append(i['name'])
            count+=1
        else:
            break
    return L
movies['cast']=movies['cast'].apply(convert3)
```

```python
movies.head()
```

```python
def find_director(obj):
    L=[]
    for i in ast.literal_eval(obj):
        if i["job"]=="Director":
            L.append(i['name'])
            break
    return L
movies['crew']=movies['crew'].apply(find_director)
```

```python
movies['genres']=movies['genres'].apply(lambda x: [i.replace(" ","") for i in x])
movies['keywords']=movies['keywords'].apply(lambda x: [i.replace(" ","") for i in x])
movies['cast']=movies['cast'].apply(lambda x: [i.replace(" ","") for i in x])
movies['crew']=movies['crew'].apply(lambda x: [i.replace(" ","") for i in x])
```

```python
movies.head()
```

```python
movies.dropna(subset=["overview"], inplace=True)
movies["overview"] = movies["overview"].apply(lambda x: x.split())
```

```python
movies.head()
```

```python
movies["tags"]=movies["overview"]+movies["genres"]+movies["keywords"]+movies["cast"]+movies["crew"]
```

```python
df=movies[["movie_id","title","tags"]]
```

```python
df.head()
```

```python
df["tags"]=df["tags"].apply(lambda x: " ".join(x))
```

```python
df["tags"]=df["tags"].apply(lambda x: x.lower())
```

```python
import nltk
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()

def stem(text):
    return " ".join([ps.stem(word) for word in text.split()])

# stem the tags column
df['tags'] = df['tags'].apply(stem)

# re-vectorize from scratch
cv = CountVectorizer(max_features=5000, stop_words='english')
vectors = cv.fit_transform(df['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(vectors)
```

```python
def recommend(movie):
    movie = movie.lower()
    matches = df[df['title'].str.lower() == movie]
    
    if matches.empty:
        print("Movie not found. Please check the spelling.")
        return
    
    movie_index = matches.index[0]
    distances = similarity[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:6]
    
    for i in movies_list:
        print(df.iloc[i[0]].title)
```
